## Download geo data

This is a tiny notebook to download the geodata necessary for the streamlit app, to not have to download it every rerun of the app.

```python
# import libraries
import pandas as pd
import geopandas as gpd


# URL for the Berlin PLR geometries
url = (
        "https://gdi.berlin.de/services/wfs/lor_2021?service=WFS&version=2.0.0"
        "&request=GetFeature&typeNames=lor_2021:a_lor_plr_2021&outputFormat=application/json"
)

plr_geo = gpd.read_file(url)
plr_geo.to_file("plr_geometries.gpkg", driver="GPKG")
```
