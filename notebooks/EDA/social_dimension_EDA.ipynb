```python
# let's create a graph for missing values

# creating a missing report first 
missing_report = (
    df_social[model_features]
    .isna()
    .agg(["sum", "mean"])
    .T
    .reset_index()
)

missing_report.columns = [
    "column_name",
    "missing_count",
    "missing_share"
]

missing_report = missing_report.sort_values(
    "missing_share",
    ascending=False
)

# to plot it

missing_to_plot = missing_report[
    missing_report["missing_count"] > 0
].copy()

if missing_to_plot.empty:
    print("No missing values in feature columns.")
else:
    fig, ax = plt.subplots(
        figsize=(10, max(4, 0.35 * len(missing_to_plot)))
    )

    ax.barh(
        missing_to_plot["column_name"],
        missing_to_plot["missing_share"],
        color='#D35FB7'
    )

    ax.set_xlabel("Share of missing values")
    ax.set_ylabel("Feature")
    ax.set_title("Missing values by feature")

    ax.invert_yaxis()

    plt.tight_layout()
    plt.show()

```

```python
# or a little nicer

def get_bins(series, max_bins=40):
    values = series.dropna()

    if values.nunique() <= 1:
        return 1

    bin_edges = np.histogram_bin_edges(values, bins="fd")
    n_bins = len(bin_edges) - 1

    return min(max(n_bins, 5), max_bins)


n_cols = 3
n_rows = int(np.ceil(len(model_features) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(18, 4 * n_rows)
)

axes = axes.flatten()

for i, col in enumerate(model_features):
    ax = axes[i]

    values = df_social[col].dropna()
    bins = get_bins(values)

    ax.hist(
        values,
        bins=bins,
        color="#40B0A6",
        edgecolor="white"
    )

    ax.axvline(
        values.mean(),
        linestyle=":",
        linewidth=1.5,
        color="black",
        label="Mean"
    )

    ax.axvline(
        values.median(),
        linestyle="--",
        linewidth=1.5,
        color="black",
        label="Median"
    )

    ax.set_title(col)
    ax.set_xlabel("Value")
    ax.set_ylabel("Number of PLR")

    if i == 0:
        ax.legend()

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.tight_layout()
plt.show()
```

```python
# creating a correlation matrix to check, if features are measuring basically the same and are therefore redundant (PEARSON CORRELATION)

plt.figure(figsize=(15, 10))
correlations = df_social[all_feature_cols].corr()
sns.heatmap(round(correlations, 3), cmap='Purples', annot=True, annot_kws={"size": 7}, vmin=-1, vmax=1)
```

```python
# let's show the highly correlated features in a table, this time we use spearman, as it is more reliable in our case (SPEARMAN CORRELATION)

threshold = 0.79

corr_spearman = df_social[all_feature_cols].corr(method="spearman")

upper_triangle = np.triu(
    np.ones(corr_spearman.shape),
    k=1
).astype(bool)

high_spearman_corr = (
    corr_spearman
    .where(upper_triangle)
    .stack()
    .reset_index()
)

high_spearman_corr.columns = [
    "feature_1",
    "feature_2",
    "spearman_correlation"
]

high_spearman_corr = high_spearman_corr[
    high_spearman_corr["spearman_correlation"].abs() > threshold
].sort_values(
    "spearman_correlation",
    key=lambda x: x.abs(),
    ascending=False
)

high_spearman_corr
```

Since the status feature of average household size is highly correlated with other variables, we will drop the other variables (households_without_minor_children_share_2024, single_person_hh_share_2024)

```python
# let's have a look at outliers

# z-score outlier detection, but this actually works best if the data is normally distributed

z_threshold = 3

zscore_outlier_report = []

for col in all_feature_cols:
    values = df_social[col]

    mean = values.mean()
    std = values.std()

    z_scores = (values - mean) / std

    outlier_count = (z_scores.abs() > z_threshold).sum()

    zscore_outlier_report.append({
        "feature": col,
        "outlier_count": outlier_count
    })

zscore_outlier_report = pd.DataFrame(zscore_outlier_report)

zscore_outlier_report.sort_values("outlier_count", ascending=False)
```

```python
# so let's also check the interquartile range, cause this is more robust regarding distribution

outlier_report = []

for col in all_feature_cols:
    values = df_social[col].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = ((values < lower_bound) | (values > upper_bound)).sum()

    outlier_report.append({
        "feature": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report.sort_values("outlier_count", ascending=False)
```
